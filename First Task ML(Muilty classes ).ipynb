```python
gar= [ 'histogram_max' ,'histogram_number_of_peaks', 'histogram_number_of_zeroes' , 'histogram_width']
df = df.drop(gar, axis=1)
```

```python
df.head(10)
```

```python
df.shape
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

new_cul =['baseline value','abnormal_short_term_variability', 'histogram_min','histogram_median','histogram_variance'
         ,'percentage_of_time_with_abnormal_long_term_variability' , 'histogram_mode' , 'histogram_mean' ]
scal_d = scaler.fit_transform(df[new_cul])
scal_df = pd.DataFrame(scal_d, columns=new_cul)
scal_df
```

```python
non_scaled_features = df.drop(columns=new_cul)
non_scaled_features.head()
```

```python
final_df = pd.concat([non_scaled_features, scal_df], axis=1)
final_df
```

```python
ax = sns.countplot(data= final_df, x="fetal_health",)
```

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state = 42)

X, y = smote.fit_resample(final_df, final_df['fetal_health'])
df_smote = pd.DataFrame(X, columns = final_df.columns)
df_smote['fetal_health'] = y
```

```python
ax = sns.countplot(data= df_smote, x="fetal_health",)

```

```python
df_smote.shape
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y,test_size=0.20, random_state=25)
```

```python
from sklearn.neighbors import KNeighborsClassifier 
knn = KNeighborsClassifier(n_neighbors=3)

knn.fit ( X_train, y_train )
```

```python
from sklearn.metrics import accuracy_score
y_pred = knn.predict(X_test)
accurcy = accuracy_score(y_test ,y_pred)
print (accurcy)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test , y_pred))
```

```python
from sklearn.svm import SVC
svm= SVC(kernel='rbf' , probability=True)
svm.fit(X_train , y_train)
```

```python
y_pred_svm = svm.predict(X_test)
accurcy = accuracy_score(y_test,y_pred_svm  )
print (accurcy)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(max_depth=20 ,random_state=42 )
dt.fit(X_train , y_train)
```

```python
y_pred = dt.predict (X_test)
accurcy = accuracy_score(y_test , y_pred)
print (accurcy)
```
